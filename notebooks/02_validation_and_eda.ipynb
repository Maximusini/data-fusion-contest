```python
import polars as pl
from datetime import datetime
```

```python
df = pl.scan_parquet('../data/train_full.parquet').filter(
    (pl.col('event_date').dt.year() == 2025) & 
    (pl.col('event_date').dt.month() == 5)
)

print(f'Всего строк в мае 2025: {df.collect().shape[0]}')
print(f'Всего мошеннических операций за май 2025: {df.group_by("target").len().collect()}')
```
